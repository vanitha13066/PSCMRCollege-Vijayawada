# Pandas DataFrames

```python
import pandas as pd
```

```python
# Creation of Dataframe using string
s = "hi hello gd evng".split()
pd.DataFrame(s)
```

```python
s1 = "how are you i am fyn".split()
pd.DataFrame(s1)
```

```python
# creation of dataframe using list
li = [[1,2,3],[4,5,6]]
pd.DataFrame(li)
```

```python
li1 = [[34,56,78],[56,77,89],[45,78,12]]
pd.DataFrame(li1)
```

```python
li2 = [[1,2,12],[56,78,90]]
pd.DataFrame(li2,index=['list1','list2'])
```

```python
# Creation of DataFrames using dictionary
d = {'key1':'h1','key2':'h2','key3':'h3','key4':'h4','key5':'h5'}
df1 = pd.DataFrame(d,index=[1,2,3,4,5])
df1
```

```python
# to delete the rows and columns- drop()
# row wise - axis=0
# column wise - axis = 1
df
```

```python
# particular row delete
df = df.drop(2,axis=0)
df
```

```python
# particular column delete
df = df.drop('Branch',axis=1)
df
```

```python
# to print specific value
df.loc[3,'name']
```

```python
# To rename the particular single column name
df.rename(columns={'rollno':'Rno'},inplace=True)
df
```

```python
# Rename multiple column names
df.columns=['NAME','ROLLNO']
df
```

```python
# to delete all row records
print(df)
```

```python
df.drop(df.index,inplace=True)
df
```

```python
# to delete entire columns
df.drop(df.columns,inplace=True,axis=1)
df
```

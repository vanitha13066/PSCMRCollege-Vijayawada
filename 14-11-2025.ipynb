```python
# import and load the data
import pandas as pd
df = pd.read_csv('employees.csv')
df
```

```python
# display the 1st 5 records
print(df.head())
```

```python
# last 5 records
print(df.tail())
```

```python
# print basic information
print(df.info())
```

```python
# print statistical summary
print(df.describe())
```

```python
# print unique departments
print(df['Department'].unique())
```

```python
# to get no of rows and columns
df.shape
```

```python
# select one column
print(df['Name'])
```

```python
# select multiple column names
print(df[['Name','Salary']])
```

```python
# Employees in the IT Department
print(df[df['Department']=='IT'])
```

```python
# Employees with salary>50000
print(df[df['Salary']>50000])
```

```python
# IT Employees in Delhi
print(df[(df['Department'] == 'IT') & (df['City'] == 'Delhi')])
```

```python
# Sort by Salary
print(df.sort_values(by='Salary', ascending=False))
```

```python
# Sort by Age
print(df.sort_values(by='Age'))
```

```python
# Add new column for 10% salary increase
df['Updated_Salary'] = df['Salary'] * 1.10
df
```

```python
# Rename 'Department' to 'Dept'
df.rename(columns={'Department': 'Dept'}, inplace=True)
```

```python
df
```

```python
# Average salary by department
print(df.groupby('Dept')['Salary'].mean())
```

```python
# Count employees per city
print(df['City'].value_counts())
```

```python
# Check for missing values
print(df.isnull().sum())
```

```python
# Save to new CSV
df.to_csv("updated_employees.csv", index=False)
print("✅ File saved as updated_employees.csv")
```

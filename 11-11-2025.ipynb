#### numpy
- Numpy means numerical python
- Numpy is a python library used for working with arrays
- It is a fundamental package for sciencific computing in python
- It is used to create multi dimensional arrays for fast operations
- Numpy arrays are stored at one continuous place in memory
- Using numpy we can perform mathematical and logical operations on arrays
  

# Advantages
- consumes less memory
- fast as compared with python lists
- convinent to use

```python
# 0-dimensional array
a = np.array(3)
print(a,type(a))
```

```python
# 1-dimensional array
a1 = np.array([1,6,8,9,4])
print(a1,type(a1))
```

```python
a1+5
```

```python
# 2- dimensional array
a3 = np.array([[5,6,8],[9,5,8]])
print(a3,type(a3))
```

```python
a3[0][1]
```

```python
# 3- dimensional array
a4 = np.array([[5,6,8],[9,5,8],[8,2,3]])
print(a4,type(a4))
```

```python
# predefined functions(or) builtin functions in numpy
print(a4)
print(a4.shape) # to get the no of rows and columns
print(a4.ndim) # to get dimensions
print(a4.size) # to get the total no of elements in array
print(a4.dtype) # to get which datatype
print(a4.itemsize) # howmany bytes - 4 bytes
```

```python
# conversion of int to complex
b = np.array([5,7,8,9],dtype='complex')
print(b,b.dtype)
```

```python
# conversion of int to float
b = np.array([5,7,8,9],dtype='float')
print(b,b.dtype)
```

```python
n = np.array([45,78,90,33,89])
print(n)
print(n.max(),n.min(),n.sum())
```

```python
# using tuple
k = np.array((56,88,66,89))
print(k,type(k))
```

```python
# using dictionary
k1 = np.array({2:89,"a":"n",9.6:90,66:89})
print(k1,type(k1))
```

```python
# range()
n =np.array(range(10))
print(n,type(n))
```

```python
n1 = np.array(range(10,51))
print(n1)
```

```python
# range with skip count
n1 = np.array(range(10,51,3))
print(n1)
```

```python
# arange()
n2 = np.array(np.arange(2.5,10.5))
print(n2)
```

```python
# linspace()
c2 = np.linspace(10,20,5)
print(c2)
```

```python
# linspace()
c2 = np.linspace(10,20,7,dtype=int)
print(c2)
```

```python
# array initialization
# zero's method
a = np.zeros(5)
print(a)
a1 = np.zeros((3,4),dtype=int)
a1
```

```python
# one's method
a = np.ones(5)
print(a)
a1 = np.ones((3,4),dtype=int)
a1
```

```python
# identity matrix
print(np.eye(5))
```

```python
print(np.eye(5,dtype=int))
```

```python
# 3,4,5 a diagonal values
np.diag([3,4,5])
```

```python
# another method to print identity matrix
n = np.identity(3,dtype=int)
print(n)
```

```python
# diagonal values
np.diag([56,89,3,7,9,3,1])
```

```python
# full() # 1-d array
print(np.full(3,7))
print(np.full(5,10))
```

```python
# 2-d array
print(np.full((2,2),3))
print(np.full((5,5),7))
```

```python
# 2-d array
li1 = [4,7,9,49]
li2 = [78,99,56,33]
b = np.array([li1,li2])
print(b)
```

```python
# arithmetic operations
n1 = np.array([[56,99,66],[89,6,44]])
print("n1 matrix is :")
print(n1)
n2 = np.array([[55,88,99],[23,76,22]])
print("n2 matrix is: ")
print(n2)
```

```python
print(n1+n2) # addition
print(n1-n2) # substraction
# dot
```

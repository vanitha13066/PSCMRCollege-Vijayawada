# Files

-  By using files we can store data permanently
# How to create a text file in python
    - by using the open() function we can create the text file
    - syntax:-    file variable_name = open('filename',modes)

# modes
----------------------------------
1.read():- 'r' -----> read or view the data in a file
2.write():- 'w' ----> we can edit,remove,store the data into a text file
3.append():- 'a' ----> we can add the data into a file

### close() 
    - syntax :-    file variable_name.close()

```python
# 1.To create the empty text file
f1 = open('data1.txt','w')
print('file created successfully')
f1.close()
```

```python
# file methods
read(),write(),append()
readline(),readlines(),seek(),tell(),split()
```

```python
f2 = open('data1.txt','w')
f2.write("hello welcome to DA Workshop")
f2.close()
print("success")
```

```python
f3 = open('data1.txt','w')
f3.write('name:vanitha'+'\n'+'rollno:345'+'\n'+'course:CSE')
f3.close()
```

```python
f4 = open('data1.txt','a')
f4.write('\nage:27')
f4.close()
```

```python
# how to read the entire data from a file
g = open('data1.txt','r')
data = g.read()
print(data)
```

```python
f4 = open('data1.txt','r')
print(f4.read(10))
print(f4.read(5))
```

```python
# with()
with open('data1.txt','r') as f2:
    print(f2.read())
```

```python
# readline()
with open('data1.txt','r') as f3:
    print(f3.readline())
```

```python
# readline()
with open('data1.txt','r') as f4:
    print(f4.readlines())
```

```python
# total no of characters in your file
with open('data1.txt','r') as f5:
    print(len(f5.read()))
```

```python
# total no of lines in your file
with open('data1.txt','r') as f5:
    print(len(f5.readlines()))
```

```python
# seek()
with open('data1.txt','r') as f4:
    print(f4.seek(5))
    print(f4.read(3))
    print(f4.read(5))
    print(f4.tell())
```

```python
with open('data1.txt','r') as h1:
    print(h1.read().split(':'))
```

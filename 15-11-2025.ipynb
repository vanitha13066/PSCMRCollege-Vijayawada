# Matplotlibrary

- It is a cross platform data visualization library for making
     2d plots to given data
  - Types of plots
        - Line plot
        - Bar graph
        - scatter plot
        - Area plot
        - Pie chart

```python
import matplotlib.pyplot as plt
# line plot
plt.plot([1,2,3,4],[5,6,7,8])
plt.title('XY Graph')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.show()
```

```python
# Bargraph is used to create bar chart
marks = [88,94,99,92,91]
names = ['C','JAVA','PYTHON','ORACLE','C++']
plt.bar(names,marks,color=['red','yellow','pink','brown','green'],width=0.6)
```

```python
# scatter plot
plt.scatter([1,2,3,4],[3,5,7,8],color='pink')
plt.show()
```

```python
# Area plot
days = [1,2,3,4]
working = [8,11,9,10]
sleeping = [8,10,11,12]
plt.stackplot(days,working,sleeping,colors=['r','g'],labels=['work','rest'])
plt.legend()
plt.show()
```

```python
# scatter plot
import numpy as np
import matplotlib.pyplot as plt
x = np.random.rand(50)
y = np.random.rand(50)
plt.scatter(x,y)
plt.title('Random Scatter')
plt.show()
```

```python
# colored scatter
import matplotlib.pyplot as plt
x = [1,2,3,4]
y = [10,22,13,40]
colors = [100,200,300,400]
plt.scatter(x,y,c=colors)
plt.title('Colored Scatter plot')
plt.show()
```

```python
# pie chart
rating = [5,3,4.5,4.8]
names = ['python','java','c++','c']
plt.pie(rating,labels=names)
plt.legend()
plt.show
```
